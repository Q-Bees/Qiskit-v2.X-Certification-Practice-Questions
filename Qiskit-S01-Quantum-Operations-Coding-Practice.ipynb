## Question 3 – Conjugation by H (single qubit)

Recall that conjugation is defined as $U P U^\dagger$.

**Q3.1**: Compute (up to global phase):
- $(H X H)$
- $(H Z H)$
- $(H Y H)$

**Q3.2**: Which single-qubit Paulis do these become? (e.g., $HXH = ?$)

_Optional: verify using a 1-qubit circuit and `Operator`._

````python
# Optional check for Q3
def conjugate_by_H(label):
    qc = QuantumCircuit(1)
    qc.h(0)
    if label == 'X':
        qc.x(0)
    elif label == 'Y':
        qc.y(0)
    elif label == 'Z':
        qc.z(0)
    qc.h(0)
    return Operator(qc)

conjugate_by_H('X'), conjugate_by_H('Z'), conjugate_by_H('Y')
````

## Question 4 – Conjugation by S

Let $S$ be the phase gate $S = \begin{pmatrix} 1 & 0 \\ 0 & i \end{pmatrix}$.

**Q4.1**: Compute (up to global phase):
- $S X S^\dagger$
- $S Y S^\dagger$
- $S Z S^\dagger$

**Q4.2**: Which Pauli does each become? (e.g., $S X S^\dagger = ?$)


_Hint: S maps X and Y into each other up to phases; Z is diagonal in the same basis as S._

````python
# Optional check for Q4
def conjugate_by_S(label):
    qc = QuantumCircuit(1)
    qc.s(0)
    if label == 'X':
        qc.x(0)
    elif label == 'Y':
        qc.y(0)
    elif label == 'Z':
        qc.z(0)
    qc.sdg(0)
    return Operator(qc)

conjugate_by_S('X'), conjugate_by_S('Y'), conjugate_by_S('Z')
````

## Question 5 – CNOT conjugation of Pauli operators

Consider a 2-qubit system with qubit 0 as control and qubit 1 as target.

Let $\text{CNOT}$ be $\text{CNOT}_{0 \to 1}$.

**Q5.1**: Under conjugation by $\text{CNOT}$, give the images of the following tensor-product Paulis (up to phase):
- $X \otimes I$
- $I \otimes X$
- $Z \otimes I$
- $I \otimes Z$

That is, compute $\text{CNOT}(P)\text{CNOT}^\dagger$ for each $P$.


_Hint: Think about how Pauli operators transform under control/target behavior of CNOT._

````python
# Optional check for Q5
def conjugate_two_qubit_pauli(pauli_label):
    # pauli_label like 'XI', 'IX', 'ZI', 'IZ'
    P = Pauli(pauli_label).to_matrix()
    qc = QuantumCircuit(2)
    qc.cx(0, 1)
    U = Operator(qc).data
    return U @ P @ U.conj().T

conjugate_two_qubit_pauli('XI'), conjugate_two_qubit_pauli('IX')
````
