**Q10**: You create the following Qiskit circuit: 

```python
qc = QuantumCircuit(1)
qc.x(0)
qc.y(0)
qc.z(0)
```
Ignoring global phase, what is the net single‑qubit operation equivalent to?

- A
Pauli‑X

- B
Identity 

- C
Pauli‑Y

- D
Pauli‑Z

**Answer**: B
Identity

````python
import numpy as np

# Define the Pauli matrices
X = np.array([[0, 1], [1, 0]])
Y = np.array([[0, -1j], [1j, 0]])
Z = np.array([[1, 0], [0, -1]])

# Calculate the sequence Z * Y * X (order of operations in circuit)
# Note: In a circuit, the first gate applied is X, then Y, then Z.
# Matrix multiplication order is Z @ Y @ X
result = Z @ Y @ X

print(f"Result matrix:\n{result}")
````
