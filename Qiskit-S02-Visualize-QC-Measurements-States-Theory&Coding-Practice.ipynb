I’ve created a 18 question intermediate to advanced level multiple-choice questions quiz with A, B, C and D options on section 2. 

Topics covered here are:

2. Visualizing Quantum States (Density Matrices & Statevectors)

- plot_state_city: Viewing the Real and Imaginary parts of the Density Matrix (

### Topic 1: Visualize Quantum Circuits

**Q1**: You have a circuit qc. Which line of code will generate a Matplotlib-based visualization of the circuit?
- A. `qc.draw()`
- B. `qc.plot()`
- C. `qc.draw(output='mpl')`
- D. `visualize_circuit(qc)`

**Answer**: C 	`output='mpl'` or `output='latex'` requires extra libraries; default is text.

**Q2**: When using `qc.draw(output='text')`, how are the qubits ordered in the output diagram?

- A. Qubit 0 is at the top, Qubit n is at the bottom.
- B. Qubit n is at the top, Qubit 0 is at the bottom.
- C. It depends on the number of classical bits.
- D. The order is randomized to optimize space.

**Answer**: A	Qubit 0 is the top-most line in the text drawer.

### Topic 2: Visualize Quantum Measurements

**Q3**: After running a simulation and getting the `counts` dictionary (e.g., `{'00': 512, '11': 512}`), which Qiskit function is used to create a bar chart of these results?

- A. `plot_state_city(counts)`
- B. `plot_histogram(counts)`
- C. `plot_bloch_multivector(counts)`
- D. `qc.plot_counts()`

**Answer**: B	`plot_histogram` is specifically for measurement counts (bitstrings).

**Q4**: You want to visualize the results of multiple different experiments on the same bar chart to compare them. Which argument should you pass to `plot_histogram()`?

- A. `legend=['Exp 1', 'Exp 2']`
- B. `multiple=True`
- C. `compare=True`
- D. `labels=['Exp 1', 'Exp 2']`

**Answer**: A	Passing a list of dictionaries to `plot_histogram()` requires the legend kwarg.

### Topic 3: Visualize Quantum States

**Q5**: Which visualization is best suited for showing the phase of individual statevector components using the color of the bars?
- A. `plot_state_hinton`
- B. `plot_state_city`
- C. `plot_state_qsphere`
- D. `plot_bloch_multivector`

**Answer**: 	C	`qsphere` uses the node color to represent the phase phi

**Q6**: Consider a 2-qubit system in the state |00>. If you use plot_bloch_multivector, how many Bloch spheres will be displayed?
- A. 1
- B. 2
- C. 4
- D. None, Bloch spheres only work for single qubits.

**Answer**: 	B	`plot_bloch_multivector` shows one sphere per qubit (by tracing out others).

**Q7**: What does plot_state_city represent?
- A. The probability of each bitstring as a 2D grid.
- B. The real and imaginary parts of the density matrix as 3D bar charts.
- C. The expectation values of X, Y, and Z.
- D. A projection of the state onto the X-Y plane.

**Answer**: 	B	"City" visualization shows the density matrix rho (Real/Imaginary).

#### Pro-Tip for the Exam:
Remember that plot_bloch_multivector and plot_state_qsphere require a Statevector or DensityMatrix object, while plot_histogram requires a Counts dictionary. Mixing them up is a common trick question!

### Advanced Questions

**Q8**: You want to draw a circuit and highlight specific gates to make the diagram easier to read. Which parameter in `qc.draw()` allows you to pass a dictionary of colors for different gate types?
- A. `colors={'h': 'red'}`
- B. `style={'displaycolor': {'h': (1, 0, 0)}}`
- C. `highlight={'h': 'red'}`
- D. `gate_colors={'h': 'red'}`

**Answer**: 	B	`qc.draw()` uses a `style` dictionary (or a pre-defined theme) for customization.

**Q9**: When using `plot_bloch_multivector(state)`, what happens if the input state is a multi-qubit entangled state (e.g., a Bell state)?
- A. It displays a single Bloch sphere representing the entire system.
- B. It displays two Bloch spheres, but both vectors will have a length of zero (pointing to the center).
- C. It throws a `VisualizationError` because entangled states cannot be split into individual vectors.
- D. It displays two Bloch spheres, each pointing to the North pole ($|00>$).

**Answer**: 	B	Important: For entangled states, the reduced density matrix of a single qubit is mixed. A maximally mixed state has a Bloch vector of length 0 (the center of the sphere).

**Q10**: You have a `Statevector` object named `sv`. You want to visualize it using `plot_state_qsphere(sv)`. What does the radius of the points (nodes) on the surface of the sphere represent?
- A. The phase of that basis state.
- B. The real part of the amplitude.
- C. The magnitude (absolute value) of the amplitude of that basis state.
- D. The probability of measuring that state in the X basis.

**Answer**: 	C	In a `qsphere`, the size of the circle is the amplitude magnitude, and the color is the phase.

**Q11**: Which of the following visualization functions is the only one that can natively accept a `DensityMatrix` object as an input?
- A. `plot_histogram`
- B. `plot_state_city`
- C. `plot_bloch_multivector`
- D. Both B and C.

**Answer**: 	D	`plot_state_city`, `plot_state_hinton`, `plot_state_paulivec`, and `plot_bloch_multivector` all accept DensityMatrix. `plot_histogram` ONLY accepts `Counts`.

**Q12**: When using the `plot_state_hinton` function, what do the size and the color (white/black or grey/white) of the squares represent?
- A. Size represents the phase; color represents the magnitude.
- B. Size represents the magnitude of the density matrix element; color represents the sign (positive/negative).
- C. Size represents the probability; color represents the qubit index.
- D. Hinton plots are only for classical bit correlations.

**Answer**: 	B	In a Hinton plot, the area of the square is proportional to the value's magnitude, and the color indicates the sign.

**Q13**: You want to save a circuit visualization to a file named `my_circuit.png` directly from the `draw()` method. Which syntax is correct?
- A. `qc.draw('mpl', filename='my_circuit.png')`
- B. `qc.draw(output='mpl').save('my_circuit.png')`
- C. `qc.draw(output='mpl', filename='my_circuit.png')`
- D. `qc.draw(output='png', path='my_circuit.png')`

**Answer**: 	C	The `filenam`e parameter is a valid kwarg in `qc.draw()` when using `mpl`, `latex`, or `latex_source`.

#### Study Tip for "Advanced" Visualization Questions:
The exam loves to ask about the Bloch Multivector vs. the QSphere.
Bloch Multivector: Good for separable (non-entangled) states. Fails to show correlations/entanglement (vectors just vanish).
QSphere: Good for entanglement. It shows the global state, and you can see how different basis states ($|00>, |01>, |10>$, etc.) relate to each other in terms of phase.

**Q14**: You are using `plot_histogram(counts)`. By default, what does the x-axis represent, and what is the order of the bitstrings?
- A. The x-axis represents the probability; bitstrings are ordered from right-to-left (Little-endian).
- B. The x-axis represents the bitstrings; they are ordered alphabetically (00, 01, 10, 11).
- C. The x-axis represents the bitstrings; they are ordered by their frequency (highest counts first).
- D. The x-axis represents the counts; bitstrings are on the y-axis.

**Answer**: 	B	`plot_histogram` orders bitstrings lexicographically (alphabetically) on the x-axis.

**Q15**: When drawing a circuit with `qc.draw(output='mpl')`, you want to change the scale (size) of the output image. Which parameter should you use?
- A. `size=2.0`
- B. `scale=1.5`
- C. `dpi=300`
- D. `figsize=(10, 5)`

**Answer**: 	B	`scale` is the correct parameter in `qc.draw()` to resize the diagram.

**Q16**: Which visualization function is specifically designed to show the expectation values of all possible Pauli strings (e.g., <XI>, <IZ>, <XX>) for a given state?
- A. `plot_state_city`
- B. `plot_state_hinton`
- C. `plot_state_paulivec`
- D. `plot_bloch_multivector`

**Answer**:  	C	`plot_state_paulivec` plots the expectation values of the Pauli operators for the state.

**Q17**: You use `plot_bloch_multivector(state)` for a single qubit. The vector is pointing directly to the right along the positive X-axis. Which state does this represent?
- A. $|0>$
- B. $|1>$
- C. $|+\rangle = \frac{|0\rangle+|1\rangle}{\sqrt{2}}$
- D. $|r\rangle = \frac{|0\rangle+i|1\rangle}{\sqrt{2}}$

**Answer**: C	The positive X-axis on the Bloch sphere corresponds to the $|+\rangle$ state

**Q18**: In `plot_state_qsphere`, what is the visual representation of a state that is a uniform superposition of all basis states with zero phase?
- A. A single large blue circle at the North pole.
- B. Many small circles of equal size and the same color distributed across the sphere.
- C. One large circle at the South pole.
- D. A line connecting the North and South poles.

**Answer**: 	B	Since it's a "uniform superposition," all amplitudes are equal (equal size), and "zero phase" means they all have the same color.


#### Quick Reference: Qiskit Visualization Functions


| Function | Best for... | Input Type | Key Visual Feature |
| :--- | :--- | :--- | :--- |
| **`plot_histogram`** | Measurement results | `Counts` (Dict) | Bar chart of bitstring frequencies. |
| **`plot_state_city`** | Density Matrix ($\rho$) | `Statevector` / `DensityMatrix` | 3D bars showing Real and Imaginary parts. |
| **`plot_state_paulivec`** | Pauli expectation values | `Statevector` / `DensityMatrix` | Shows $\langle X \rangle, \langle Y \rangle, \langle Z \rangle$ etc. |
| **`plot_state_qsphere`** | Entanglement & Phases | `Statevector` / `DensityMatrix` | Node **size** is amplitude, **color** is phase. |
| **`plot_bloch_multivector`** | Individual qubit states | `Statevector` / `DensityMatrix` | One sphere per qubit (fails for entanglement). |
| **`plot_state_hinton`** | Density Matrix elements | `Statevector` / `DensityMatrix` | 2D grid where square size is magnitude. |
