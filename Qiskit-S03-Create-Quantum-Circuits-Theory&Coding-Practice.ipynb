## Advanced Circuit Composition
**Q5**: You have two circuits, qc_a (qubits 0, 1) and qc_b (qubits 0, 1). You want to merge them such that qc_b is appended to the end of qc_a without creating a nested "instruction" block. Which method is most appropriate?
Options:

* A) qc_a.append(qc_b)
* B) qc_a.compose(qc_b)
* C) qc_a + qc_b
* D) qc_a.add(qc_b)

**Answer**: B. 
The compose() method merges the instructions of one circuit into another. While + was used in older versions of Qiskit, it is now deprecated or removed in favor of compose(). append() would treat qc_b as a single composite gate/instruction.

## Circuit Depth & Barriers
**Q8**: Consider a circuit where qubit 0 has an $H$ gate, then a barrier, and then another $H$ gate. What is the depth of this circuit?
Options:

* A) 1
* B) 2
* C) 3
* D) 0

**Answer**: B. 
Barriers are not counted when calculating circuit depth. The depth is determined by the longest path of non-idle gates, so two $H$ gates in sequence result in a depth of 2.

## Complex Parameter Binding with ParameterVector
**Q9**: You have a QuantumCircuit with 5 qubits and a ParameterVector named params of length 10. You want to bind the first 5 parameters to the value $0$ and the remaining 5 to $\pi$. Which approach is valid and efficient?
Options:

* A) qc.assign_parameters({params: [0]*5 + [np.pi]*5})
* B) qc.bind_parameters([0, 0, 0, 0, 0, np.pi, np.pi, np.pi, np.pi, np.pi])
* C) for i in range(5): qc.assign_parameters({params[i]: 0, params[i+5]: np.pi}, inplace=True)
* D) All of the above.

**Answer**: D. 
Qiskit is flexible with parameter assignment. assign_parameters can take a dictionary mapping a ParameterVector directly to a list of values (Option A). Option B works but relies on the internal alphabetical/numerical sorting of parameters. Option C is a manual iterative approach that also works using the inplace=True flag.

## Nested Control Flow Logic
**Q10**: You are building a "repeat-until-success" circuit. You need to apply a subcircuit trial_circ and then check a 2-bit classical register creg. If creg is not equal to 00, you must reset the qubits and repeat the trial. Since Qiskit does not support a direct "not equal" condition in if_test, how should you implement this?
Options:

* A) with qc.if_test((creg, 0b00)): pass; else_: # reset and repeat logic
* B) qc.while_loop((creg, 0b01), trial_circ)
* C) with qc.if_test((creg, 1, 2)): # reset logic
* D) qc.c_if(creg, !=0)

**Answer**: A. Current Qiskit hardware and software limitations often require using an empty if block with an else_ branch to simulate "if not equal to" logic. while_loop is also a valid structure for repetition, but the condition must be a specific value or a boolean expression, and 0b01 would only catch one of the three "failure" states.
