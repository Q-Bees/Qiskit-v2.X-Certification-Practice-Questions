# Qiskit Certification – Section 7 Theory & Coding Practice

Section 7: **Retrieve and analyze the results of quantum circuits**

Topics covered here:
- Retrieve previous experiment results (session/runtime)
- Monitor jobs

The quiz has 15 multiple choice format questions with A, B, C and D options, to test conceptual understanding and code-related knowledge. It covers key fundamental concepts such as the theoretical background of previous experiment results from sessions/runtimes, monitoring jobs, saving/retrieving jobs, and monitoring or canceling jobs. It also covers practical Qiskit Runtime features, like job management APIs and result handling.

There are also 10 coding questions covering key topics like QiskitRuntimeService.jobs(), job.status(), job.result(), job.cancel(), and session handling.
